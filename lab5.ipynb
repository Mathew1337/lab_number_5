```
class TreeNode:
    """Узел бинарного дерева поиска"""
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


def insert(root, value):
    """
    Вставить значение в дерево поиска

    Args:
        root: корень дерева (может быть None)
        value: целое число для вставки

    Returns:
        корень дерева (может измениться при первой вставке)
    """
    if root is None:
        return TreeNode(value)

    if value < root.value:
        root.left = insert(root.left, value)
    elif value > root.value:
        root.right = insert(root.right, value)
    # Если значение уже существует, ничего не делаем

    return root


def search(root, target):
    """
    Найти значение в дереве

    Args:
        root: корень дерева
        target: искомое значение

    Returns:
        True, если значение найдено, иначе False
    """
    if root is None:
        return False

    if target == root.value:
        return True
    elif target < root.value:
        return search(root.left, target)
    else:
        return search(root.right, target)


def inorder(root):
    """
    Симметричный обход (вывод в порядке возрастания)

    Args:
        root: корень дерева
    """
    if root is not None:
        inorder(root.left)
        print(root.value, end=" ")
        inorder(root.right)


def height(root):
    """
    Вычислить высоту дерева (максимальную глубину)

    Args:
        root: корень дерева

    Returns:
        высоту дерева (количество рёбер от корня до самого дальнего листа)
        для пустого дерева возвращает -1
    """
    if root is None:
        return -1

    left_height = height(root.left)
    right_height = height(root.right)

    return max(left_height, right_height) + 1


# Альтернативная версия height (если считать высоту как количество узлов)
def height_nodes(root):
    """
    Вычислить высоту дерева (количество узлов на самом длинном пути)

    Args:
        root: корень дерева

    Returns:
        количество узлов на самом длинном пути
        для пустого дерева возвращает 0
    """
    if root is None:
        return 0

    return max(height_nodes(root.left), height_nodes(root.right)) + 1


# Демонстрация работы
if __name__ == "__main__":
    # Создаём дерево
    root = None
    values = [50, 30, 70, 20, 40, 60, 80, 35, 45]

    print("Вставляем значения:", values)
    for val in values:
        root = insert(root, val)

    # Вывод в порядке возрастания
    print("\nСимметричный обход (inorder):")
    inorder(root)
    print()

    # Поиск значений
    print("\nПоиск значений:")
    for target in [40, 25, 70, 100]:
        found = search(root, target)
        print(f"Значение {target}: {'найдено' if found else 'не найдено'}")

    # Вычисление высоты
    print(f"\nВысота дерева (по рёбрам): {height(root)}")
    print(f"Высота дерева (по узлам): {height_nodes(root)}")

    # Дополнительная демонстрация с пустым деревом
    empty_root = None
    print(f"\nВысота пустого дерева (по рёбрам): {height(empty_root)}")
    print(f"Высота пустого дерева (по узлам): {height_nodes(empty_root)}")

    # Дерево из одного узла
    single_node = TreeNode(10)
    print(f"\nВысота дерева из одного узла (по рёбрам): {height(single_node)}")
    print(f"Высота дерева из одного узла (по узлам): {height_nodes(single_node)}")
```
